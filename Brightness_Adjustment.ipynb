```python
# ==============================
# 1. IMPORT LIBRARY
# ==============================

import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

library digunakan untuk mendukung proses Brightness Adjustment dan visualisasi citra digital.

- `import cv2` mengimpor OpenCV untuk membaca dan mendukung pengolahan gambar.
- `import numpy as np` mengimpor NumPy dengan alias `np` untuk melakukan perhitungan numerik pada nilai piksel.
- `import matplotlib.pyplot as plt` mengimpor Matplotlib dengan alias `plt` untuk menampilkan gambar dan hasil pemrosesan.
- `import os` mengimpor modul OS untuk mengakses informasi file, seperti ukuran gambar.
- `import` digunakan untuk memuat library atau modul agar fungsinya dapat digunakan dalam program, sedangkan `as` memberikan nama alias agar penulisan kode lebih singkat.

```python
# ==============================
# 2. MEMBACA GAMBAR ASLI
# ==============================

# Menentukan lokasi file gambar
lokasi_gambar = "foto asli.jpg"

# Membaca gambar menggunakan OpenCV
gambar_asli = cv2.imread(lokasi_gambar)

# Memeriksa apakah gambar berhasil dibaca
if gambar_asli is None:
    print("Gambar gagal dibaca. Periksa nama dan lokasi file.")
else:
    print("Gambar berhasil dibaca.")
```

Gambar asli dibaca dari file untuk digunakan dalam proses pengolahan citra digital.

- `lokasi_gambar = "foto asli.jpg"` menentukan nama file gambar yang akan dibaca. Variabel `lokasi_gambar` menyimpan nama file agar dapat digunakan pada proses berikutnya.

- `gambar_asli = cv2.imread(lokasi_gambar)` membaca gambar menggunakan OpenCV. Data piksel gambar yang berhasil dibaca disimpan dalam variabel `gambar_asli` dalam bentuk array numerik.

- `if gambar_asli is None:` memeriksa apakah gambar gagal dibaca. Nilai `None` menunjukkan bahwa OpenCV tidak berhasil memuat data gambar.

- `print("Gambar gagal dibaca. Periksa nama dan lokasi file.")` menampilkan pesan jika gambar tidak berhasil dibaca, sehingga nama dan lokasi file dapat diperiksa kembali.

- `else:` menjalankan perintah berikutnya jika kondisi pada `if` tidak terpenuhi, yang berarti gambar berhasil dibaca.

- `print("Gambar berhasil dibaca.")` menampilkan pesan bahwa gambar berhasil dimuat dan siap digunakan pada tahap selanjutnya.

```python
# ==============================
# 3. MENAMPILKAN GAMBAR ASLI
# ==============================

# Mengubah format warna BGR menjadi RGB
gambar_rgb = cv2.cvtColor(gambar_asli, cv2.COLOR_BGR2RGB)

# Mengatur ukuran visualisasi gambar
plt.figure(figsize=(5, 6))

# Menampilkan gambar asli
plt.imshow(gambar_rgb)
plt.title("Gambar Asli")
plt.axis("off")

# Merapikan tata letak visualisasi
plt.tight_layout()
plt.show()
```

Gambar asli ditampilkan untuk melihat citra yang akan digunakan dalam proses Brightness Adjustment sebelum dilakukan perubahan kecerahan.

- `gambar_rgb = cv2.cvtColor(gambar_asli, cv2.COLOR_BGR2RGB)` mengubah urutan kanal warna gambar dari BGR menjadi RGB. OpenCV menggunakan urutan BGR, sedangkan Matplotlib menggunakan RGB. Variabel `gambar_rgb` menyimpan hasil konversi agar warna gambar tampil dengan benar.

- `plt.figure(figsize=(5, 6))` membuat area visualisasi dengan lebar 5 inci dan tinggi 6 inci agar ukuran gambar tidak terlalu besar.

- `plt.imshow(gambar_rgb)` menampilkan gambar berdasarkan data piksel yang tersimpan dalam variabel `gambar_rgb`.

- `plt.title("Gambar Asli")` memberikan judul "Gambar Asli" di atas visualisasi agar gambar mudah dikenali.

- `plt.axis("off")` menyembunyikan sumbu beserta angka koordinatnya agar tampilan gambar lebih bersih.

- `plt.tight_layout()` mengatur tata letak visualisasi agar elemen gambar dan judul tersusun dengan rapi.

- `plt.show()` menampilkan hasil visualisasi gambar pada output Jupyter Notebook

```python
# ==============================
# 4. INFORMASI GAMBAR ASLI
# ==============================

# Mengambil dimensi dan jumlah kanal gambar
tinggi, lebar, kanal = gambar_asli.shape

# Mengambil tipe data piksel
tipe_data = gambar_asli.dtype

# Menghitung jumlah seluruh piksel
jumlah_piksel = tinggi * lebar

# Menghitung ukuran file dalam KB
ukuran_file = os.path.getsize(lokasi_gambar) / 1024

# Menghitung nilai intensitas piksel
nilai_minimum = np.min(gambar_asli)
nilai_maksimum = np.max(gambar_asli)
nilai_rata_rata = np.mean(gambar_asli)

# Menyusun informasi gambar
informasi_gambar = [
    ["Nama file", lokasi_gambar],
    ["Sumber gambar", "Pinterest"],
    ["Format file", lokasi_gambar.split(".")[-1].upper()],
    ["Lebar gambar", f"{lebar} piksel"],
    ["Tinggi gambar", f"{tinggi} piksel"],
    ["Resolusi spasial", f"{lebar} x {tinggi} piksel"],
    ["Jumlah piksel", f"{jumlah_piksel:,} piksel"],
    ["Jumlah kanal warna", str(kanal)],
    ["Urutan kanal", "BGR (OpenCV)"],
    ["Tipe data piksel", str(tipe_data)],
    ["Ukuran file", f"{ukuran_file:.2f} KB"],
    ["Intensitas minimum", str(nilai_minimum)],
    ["Intensitas maksimum", str(nilai_maksimum)],
    ["Rata-rata intensitas", f"{nilai_rata_rata:.2f}"],
]

# Menampilkan informasi dalam bentuk teks dan tabel sederhana
print("INFORMASI GAMBAR ASLI")
print("=" * 48)
print(f"{'Informasi':<25} | {'Nilai'}")
print("-" * 48)

for informasi, nilai in informasi_gambar:
    print(f"{informasi:<25} | {nilai}")

print("=" * 48)
```

Informasi gambar asli digunakan untuk mengetahui karakteristik gambar sebelum proses Brightness Adjustment dilakukan. Informasi ini mencakup dimensi, jumlah piksel, ukuran file, serta statistik nilai intensitas piksel.

- `gambar_asli.shape` mengambil tinggi, lebar, dan jumlah kanal warna gambar. Informasi ini digunakan untuk mengetahui dimensi gambar dan jumlah kanal yang menyimpan informasi warna.

- `gambar_asli.dtype` menunjukkan tipe data piksel. Pada gambar 8-bit, tipe data umumnya `uint8`, dengan rentang nilai 0–255. Nilai 0 menunjukkan intensitas paling rendah, sedangkan 255 menunjukkan intensitas paling tinggi.

- `tinggi * lebar` menghitung jumlah seluruh piksel berdasarkan tinggi dan lebar gambar. Jumlah ini tidak dikalikan dengan jumlah kanal karena yang dihitung adalah piksel, bukan seluruh nilai kanal.

- `os.path.getsize(lokasi_gambar) / 1024` mengambil ukuran file dalam byte, kemudian membaginya dengan 1024 untuk memperoleh ukuran dalam KB.

- `np.min(gambar_asli)` mencari nilai intensitas terkecil dari seluruh kanal piksel. Nilai ini menunjukkan intensitas terendah yang ditemukan pada gambar.

- `np.max(gambar_asli)` mencari nilai intensitas terbesar dari seluruh kanal piksel. Nilai ini menunjukkan intensitas tertinggi yang ditemukan pada gambar.

- `np.mean(gambar_asli)` menghitung rata-rata seluruh nilai kanal piksel dengan menjumlahkan nilai-nilai tersebut, kemudian membaginya dengan jumlah nilai yang dihitung. Rata-rata ini memberikan gambaran umum intensitas gambar, tetapi tidak sepenuhnya mewakili persepsi kecerahan manusia.

- `informasi_gambar` menyimpan informasi gambar dalam bentuk daftar yang berisi nama informasi dan nilainya, seperti nama file, sumber gambar, format, dimensi, jumlah piksel, kanal warna, tipe data, ukuran file, dan statistik intensitas.

- `print()` menampilkan informasi dalam bentuk teks, sedangkan pengaturan lebar kolom dan garis pemisah membantu menyusun hasil agar mudah dibaca.

- `for informasi, nilai in informasi_gambar` membaca setiap pasangan informasi dan nilai dari daftar, kemudian menampilkannya satu per satu.

Nilai minimum, maksimum, dan rata-rata nantinya dapat dibandingkan dengan hasil setelah Brightness Adjustment. Perbandingan tersebut membantu menjelaskan perubahan intensitas piksel berdasarkan data numerik, bukan hanya berdasarkan tampilan gambar.

```python
# ==============================
# 5. MENENTUKAN NILAI KECERAHAN
# ==============================

# Nilai positif untuk mencerahkan gambar
nilai_terang = 50

# Nilai negatif untuk menggelapkan gambar
nilai_gelap = -50

# Menampilkan nilai perubahan kecerahan
print("Nilai perubahan kecerahan")
print(f"Gambar lebih terang : {nilai_terang}")
print(f"Gambar lebih gelap   : {nilai_gelap}")
```

Nilai perubahan kecerahan ditentukan untuk membuat gambar lebih terang dan lebih gelap.

- `nilai_terang = 50` menambahkan intensitas piksel agar gambar lebih terang.

- `nilai_gelap = -50` mengurangi intensitas piksel agar gambar lebih gelap.

- `print()` menampilkan nilai perubahan kecerahan, sedangkan `f` dan `{}` digunakan untuk menampilkan nilai variabel di dalam teks.

Nilai positif meningkatkan kecerahan, sedangkan nilai negatif menurunkannya.

```python
# ==============================
# 6. MELAKUKAN BRIGHTNESS ADJUSTMENT
# ==============================

# Membuat gambar lebih terang
gambar_terang = np.clip(
    gambar_asli.astype(np.int16) + nilai_terang, 0, 255
).astype(np.uint8)

# Membuat gambar lebih gelap
gambar_gelap = np.clip(
    gambar_asli.astype(np.int16) + nilai_gelap, 0, 255
).astype(np.uint8)

print("Proses Brightness Adjustment berhasil.")
```

Pada tahap ini, gambar asli diolah untuk menghasilkan gambar yang lebih terang dan lebih gelap.

- `gambar_asli.astype(np.int16)` mengubah tipe data sementara agar perhitungan intensitas aman.

- `np.clip(..., 0, 255)` menjaga nilai piksel tetap dalam rentang 0–255.

- `nilai_terang` dan `nilai_gelap` digunakan untuk menaikkan atau menurunkan intensitas piksel.

- `.astype(np.uint8)` mengembalikan hasil ke tipe data piksel 8-bit.

```python
# ==============================
# 7. MENAMPILKAN PERBANDINGAN GAMBAR
# ==============================

# Mengubah warna BGR menjadi RGB
gambar_asli_rgb = cv2.cvtColor(gambar_asli, cv2.COLOR_BGR2RGB)
gambar_terang_rgb = cv2.cvtColor(gambar_terang, cv2.COLOR_BGR2RGB)
gambar_gelap_rgb = cv2.cvtColor(gambar_gelap, cv2.COLOR_BGR2RGB)

# Membuat area untuk tiga gambar
plt.figure(figsize=(12, 5))

# Menampilkan gambar asli
plt.subplot(1, 3, 1)
plt.imshow(gambar_asli_rgb)
plt.title("Gambar Asli")
plt.axis("off")

# Menampilkan gambar lebih terang
plt.subplot(1, 3, 2)
plt.imshow(gambar_terang_rgb)
plt.title("Lebih Terang (+50)")
plt.axis("off")

# Menampilkan gambar lebih gelap
plt.subplot(1, 3, 3)
plt.imshow(gambar_gelap_rgb)
plt.title("Lebih Gelap (-50)")
plt.axis("off")

# Merapikan dan menampilkan hasil
plt.tight_layout()
plt.show()
```

Tahap ini menampilkan gambar asli, gambar yang lebih terang, dan gambar yang lebih gelap secara berdampingan agar perbedaan hasil Brightness Adjustment dapat diamati.

- `cv2.cvtColor(gambar_asli, cv2.COLOR_BGR2RGB)` mengubah urutan kanal warna gambar asli dari BGR menjadi RGB. Konversi yang sama dilakukan pada `gambar_terang` dan `gambar_gelap` agar warna ketiga gambar tampil dengan benar menggunakan Matplotlib.

- `plt.figure(figsize=(12, 5))` membuat area visualisasi dengan lebar 12 inci dan tinggi 5 inci agar ketiga gambar dapat ditampilkan dalam satu baris.

- `plt.subplot(1, 3, 1)` membagi area visualisasi menjadi satu baris dan tiga kolom, kemudian memilih posisi pertama untuk gambar asli.

- `plt.imshow(gambar_asli_rgb)` menampilkan gambar asli, sedangkan `plt.title("Gambar Asli")` memberikan judul pada gambar.

- `plt.axis("off")` menyembunyikan sumbu dan angka koordinat agar tampilan gambar lebih rapi.

- `plt.subplot(1, 3, 2)` memilih posisi kedua untuk gambar yang lebih terang. `plt.imshow(gambar_terang_rgb)` menampilkan hasilnya dan `plt.title("Lebih Terang (+50)")` menunjukkan nilai penambahan kecerahan.

- `plt.subplot(1, 3, 3)` memilih posisi ketiga untuk gambar yang lebih gelap. `plt.imshow(gambar_gelap_rgb)` menampilkan hasilnya dan `plt.title("Lebih Gelap (-50)")` menunjukkan nilai pengurangan kecerahan.

- `plt.tight_layout()` mengatur jarak antargambar dan judul agar tidak bertumpuk.

- `plt.show()` menampilkan seluruh visualisasi.

Melalui perbandingan ini, dapat diamati bahwa penambahan intensitas membuat gambar lebih terang, sedangkan pengurangan intensitas membuat gambar lebih gelap. Perubahan dilakukan pada nilai piksel, bukan dengan mengubah ukuran gambar.

```python
# ==============================
# 8. MENYIMPAN HASIL PENGOLAHAN CITRA
# ==============================

# Menentukan nama file hasil
file_terang = "hasil/gambar_terang.jpg"
file_gelap = "hasil/gambar_gelap.jpg"

# Menyimpan gambar hasil Brightness Adjustment
cv2.imwrite(file_terang, gambar_terang)
cv2.imwrite(file_gelap, gambar_gelap)

print("Gambar hasil berhasil disimpan.")
print("Lokasi:", file_terang, "dan", file_gelap)
```

Tahap ini digunakan untuk menyimpan gambar hasil Brightness Adjustment ke dalam folder `hasil`, sehingga gambar yang telah diproses dapat dibuka kembali tanpa harus menjalankan proses dari awal.

- `file_terang = "hasil/gambar_terang.jpg"` menentukan lokasi dan nama file untuk menyimpan gambar yang lebih terang. Folder `hasil` merupakan tempat penyimpanan, sedangkan `gambar_terang.jpg` adalah nama file hasilnya.

- `file_gelap = "hasil/gambar_gelap.jpg"` menentukan lokasi dan nama file untuk menyimpan gambar yang lebih gelap secara terpisah dari gambar terang.

- `cv2.imwrite(file_terang, gambar_terang)` menyimpan data piksel dari variabel `gambar_terang` ke dalam file JPG sesuai lokasi yang telah ditentukan.

- `cv2.imwrite(file_gelap, gambar_gelap)` menyimpan data piksel dari variabel `gambar_gelap` ke dalam file JPG. Dengan demikian, kedua hasil pengolahan memiliki file masing-masing.

- `print("Gambar hasil berhasil disimpan.")` menampilkan pesan bahwa proses penyimpanan telah dijalankan.

- `print("Lokasi:", file_terang, "dan", file_gelap)` menampilkan lokasi dan nama kedua file agar hasil penyimpanan mudah diketahui.

Hasilnya adalah dua file terpisah, yaitu `gambar_terang.jpg` dan `gambar_gelap.jpg`, yang dapat dilihat pada folder `hasil` di Explorer VS Code. Gambar asli tetap tersimpan dan tidak ditimpa oleh hasil pengolahan.

```python
# ==============================
# 9. ANALISIS NILAI INTENSITAS PIKSEL
# ==============================

# Menghitung statistik intensitas setiap gambar
gambar_dianalisis = [
    ("Gambar Asli", gambar_asli),
    ("Gambar Terang", gambar_terang),
    ("Gambar Gelap", gambar_gelap)
]

print("ANALISIS INTENSITAS PIXEL")
print("=" * 55)
print(f"{'Gambar':<18} | {'Minimum':>8} | {'Maksimum':>8} | {'Rata-rata':>9}")
print("-" * 55)

for nama, gambar in gambar_dianalisis:
    minimum = np.min(gambar)
    maksimum = np.max(gambar)
    rata_rata = np.mean(gambar)

    print(f"{nama:<18} | {minimum:>8} | {maksimum:>8} | {rata_rata:>9.2f}")

print("=" * 55)
```

Analisis ini dilakukan untuk membandingkan nilai intensitas piksel pada gambar asli, gambar terang, dan gambar gelap secara numerik.

- `gambar_dianalisis` menyimpan nama dan data dari ketiga gambar dalam bentuk daftar agar dapat dianalisis secara berurutan.

- `np.min(gambar)` menghitung nilai intensitas piksel paling rendah pada setiap gambar.

- `np.max(gambar)` menghitung nilai intensitas piksel paling tinggi pada setiap gambar.

- `np.mean(gambar)` menghitung rata-rata seluruh nilai intensitas kanal piksel untuk mengetahui perubahan intensitas secara umum.

- `for nama, gambar in gambar_dianalisis:` mengulang proses analisis untuk setiap gambar, sehingga perhitungan tidak perlu ditulis berulang kali.

- `print()` menampilkan hasil analisis dalam bentuk tabel teks. Pengaturan seperti `<18`, `>8`, dan `>9.2f` mengatur perataan serta lebar kolom agar hasilnya rapi. `.2f` menampilkan rata-rata dengan dua angka di belakang koma.

Berdasarkan hasil pengujian, rata-rata intensitas gambar asli adalah **63,58**, gambar terang meningkat menjadi **113,45**, sedangkan gambar gelap menurun menjadi **18,20**. Hasil ini membuktikan bahwa penambahan nilai kecerahan meningkatkan intensitas piksel, sementara pengurangan nilai kecerahan menurunkannya.

Nilai minimum dan maksimum juga dipengaruhi oleh pembatasan intensitas pada rentang 0–255. Karena itu, perubahan kecerahan tidak selalu membuat nilai minimum atau maksimum berubah dengan besaran yang sama.
